```python
%%writefile arquitectura_tomato.puml
@startuml
title Arquitectura del Sistema Inteligente de Detección de Enfermedades del Tomate

skinparam componentStyle rectangle
skinparam shadowing false

actor Usuario

package "Cliente" {
    [Frontend Web\nHTML/CSS/JavaScript]
}

package "Servidor" {
    [API Flask]
    [Preprocesamiento\n(OpenCV + Pillow)]
    [Modelo CNN\nResNet18 - PyTorch]
    [Grad-CAM]
}

database "Base de Datos\nEnfermedades" as BD

Usuario --> [Frontend Web\nHTML/CSS/JavaScript] : Selecciona imagen

[Frontend Web\nHTML/CSS/JavaScript] --> [API Flask] : POST /predict

[API Flask] --> [Preprocesamiento\n(OpenCV + Pillow)] : Procesar imagen

[Preprocesamiento\n(OpenCV + Pillow)] --> [Modelo CNN\nResNet18 - PyTorch] : Imagen normalizada

[Modelo CNN\nResNet18 - PyTorch] --> [Grad-CAM] : Activaciones

[Modelo CNN\nResNet18 - PyTorch] --> BD : Consultar enfermedad

BD --> [API Flask] : Síntomas y tratamiento

[Grad-CAM] --> [API Flask] : Mapa de calor

[API Flask] --> [Frontend Web\nHTML/CSS/JavaScript] : JSON + GradCAM

[Frontend Web\nHTML/CSS/JavaScript] --> Usuario : Mostrar diagnóstico

@enduml
```

```python
%%writefile actividades_tomato.puml
@startuml
title Diagrama de Actividades - Detección de Enfermedades del Tomate

start

:Ingresar al sistema;

:Seleccionar imagen;

if (¿Imagen válida?) then (Sí)

:Preprocesar imagen;

:Cargar modelo CNN;

:Clasificar enfermedad;

:Generar Grad-CAM;

:Consultar síntomas y tratamiento;

:Mostrar diagnóstico;

else (No)

:Mostrar mensaje de error;

endif

stop

@enduml
```

```python
!java -jar plantuml.jar actividades_tomato.puml
```

```python
%%writefile componentes_tomato.puml
@startuml
title Diagrama de Componentes

package "Frontend" {
component "HTML"
component "CSS"
component "JavaScript"
}

package "Backend Flask" {
component "API REST"
component "Preprocesamiento"
component "CNN ResNet18"
component "GradCAM"
}

database "Base de Datos"

HTML --> "API REST"
JavaScript --> "API REST"

"API REST" --> Preprocesamiento
Preprocesamiento --> "CNN ResNet18"
"CNN ResNet18" --> GradCAM
"CNN ResNet18" --> "Base de Datos"

@enduml
```

```python
%%writefile entrenamiento_tomato.puml
@startuml
title Entrenamiento del Modelo CNN

start

:Dataset PlantVillage;

:Dividir Train/Test;

:Aplicar Data Augmentation;

:Entrenar ResNet18;

:Validar Modelo;

:Calcular Accuracy;

:Guardar tomato_model.pth;

stop

@enduml
```

```python
%%writefile secuencia_tomato.puml
@startuml
title Diagrama de Secuencia

actor Usuario

participant Frontend
participant Flask
participant CNN
participant GradCAM

Usuario -> Frontend : Subir imagen
Frontend -> Flask : POST /predict
Flask -> CNN : Clasificar imagen
CNN --> Flask : Resultado
Flask -> GradCAM : Generar mapa
GradCAM --> Flask : Imagen
Flask --> Frontend : JSON
Frontend --> Usuario : Mostrar diagnóstico

@enduml
```

```python
%%writefile clases_tomato.puml
@startuml
title Diagrama de Clases

class Usuario{
+id
+nombre
}

class Prediccion{
+imagen
+resultado
+confianza
}

class Enfermedad{
+nombre
+sintomas
+tratamiento
}

class ModeloCNN{
+cargarModelo()
+predecir()
+generarGradCAM()
}

Usuario --> Prediccion
Prediccion --> Enfermedad
Prediccion --> ModeloCNN

@enduml
```

```python
%%writefile README.md

# Tomato Disease ML 🍅🤖

Sistema inteligente para detectar enfermedades del tomate mediante visión artificial.

## Tecnologías

- Python
- PyTorch
- ResNet18 CNN
- Flask
- HTML/CSS/JavaScript
- Dataset PlantVillage

## Funcionamiento

1. Usuario carga una imagen del tomate.
2. El sistema realiza preprocesamiento.
3. La CNN analiza la imagen.
4. Se obtiene la enfermedad detectada.
5. Se muestra el resultado.

## Modelo

Red neuronal convolucional entrenada con imágenes de hojas de tomate.

## Documentación UML

Incluye:

- Arquitectura
- Componentes
- Clases
- Secuencia
- Actividades
- Entrenamiento
- Despliegue
```
