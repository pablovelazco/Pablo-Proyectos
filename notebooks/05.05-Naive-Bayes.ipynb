```python
import networkx as nx
import matplotlib.pyplot as plt

# Crear red bayesiana
red = nx.DiGraph()

# Tres variables
red.add_nodes_from([
    "Lluvia",
    "Suelo mojado",
    "Humedad alta"
])

# Relaciones
red.add_edge("Lluvia", "Suelo mojado")
red.add_edge("Lluvia", "Humedad alta")
red.add_edge("Suelo mojado", "Humedad alta")

# Posiciones
pos = {
    "Lluvia": (0, 1),
    "Suelo mojado": (-1, 0),
    "Humedad alta": (1, 0)
}

# IMAGEN MÁS PEQUEÑA
plt.figure(figsize=(5, 3.5))

nx.draw(
    red,
    pos,
    with_labels=True,
    node_size=2800,
    font_size=9,
    font_weight="bold",
    arrows=True,
    arrowsize=18
)

plt.title("Red Bayesiana de 3 variables", fontsize=11)

plt.tight_layout()
plt.show()
```
